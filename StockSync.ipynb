## 5. Unified Analytics Dashboard Implementation
The function block below generates a dual-axis financial dashboard. It contrasts daily market valuation vectors with quarterly corporate earnings indicators.


```python
# The make_graph function has been modified to use Matplotlib for static graphs. Earlier, it used Plotly to generate interactive dashboards, which caused issues when uploading the notebook in the MARK assignment submission.



import matplotlib.pyplot as plt

def make_graph(stock_data, revenue_data, stock):
    stock_data_specific = stock_data[stock_data.Date <= '2026-06-30']
    revenue_data_specific = revenue_data[revenue_data.Date <= '2026-06-30']

    fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

    # Stock price
    axes[0].plot(pd.to_datetime(stock_data_specific.Date), stock_data_specific.Close.astype("float"), label="Share Price", color="blue")
    axes[0].set_ylabel("Price ($US)")
    axes[0].set_title(f"{stock} - Historical Share Price")

    # Revenue
    axes[1].plot(pd.to_datetime(revenue_data_specific.Date), revenue_data_specific.Revenue.astype("float"), label="Revenue", color="green")
    axes[1].set_ylabel("Revenue ($US Millions)")
    axes[1].set_xlabel("Date")
    axes[1].set_title(f"{stock} - Historical Revenue")

    plt.tight_layout()
    plt.savefig('stock_chart.png')
    plt.show()
```

The presentation template compiled above will serve as the terminal mapping destination for both target assets processed downstream.

## 6. Module I: Programmatic Ingestion Framework (Asset: NVIDIA / NVDA)
Programmatic historical index tracking parameters are mapped below to generate a native structured dataframe wrapper.


```python
nvdia = yf.Ticker("NVDA")
nvdia_data = nvdia.history(period = "max")
```

### 6.1 Index Refactoring & Head Vector Check
Structural row index keys are flattened into explicit column variables before inspecting the initial tracking elements.


```python
nvdia_data.reset_index(inplace = True)
nvdia_data.head()
```

## 7. Module II: Document Object Model Scraping Pipeline (Asset A Fundamentals Ingestion)
The web harvester utilizes remote requests to clone unstructured financial reporting matrices before loading them into parsing buffers.


```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/revenue.htm"
html_data = requests.get(url).text
```

### 7.1 Structural Ingestion Parsing
The target raw stream payload is structured into an active searchable DOM tree object.


```python
soup = BeautifulSoup(html_data, "html.parser")
```

### 7.2 Matrix Traversal & Table Data Array Loading
The parsing block below iterates over targeted document tree rows to parse and extract string values into a relational framework.


```python
nvdia_revenue = pd.DataFrame(columns = ["Date","Revenue"])
for row in soup.find_all("tbody")[1].find_all("tr"):
    col = row.find_all("td")
    date = col[0].text
    rev = col[1].text
    nvdia_revenue = pd.concat([nvdia_revenue, pd.DataFrame({"Date" : [date], "Revenue" : [rev]})], ignore_index = True)
```

```python
nvdia_revenue["Revenue"] = nvdia_revenue['Revenue'].str.replace(',|\$',"",regex=True)
```

### 7.4 Structural Continuity Checks
Null arrays and whitespace strings are dropped from the operational dataset.

```python
nvdia_revenue.dropna(inplace=True)

nvdia_revenue = nvdia_revenue[nvdia_revenue['Revenue'] != ""]
```

### 7.5 Base Boundary Matrix Review
The terminal collection lines are rendered below to confirm successful payload structural conversions.


```python
nvdia_revenue.tail()
```

## 8. Module III: Programmatic Ingestion Framework (Asset B - Microsoft)

```python
microsoft = yf.Ticker("MSFT")
microsoft_data = microsoft.history(period = "max")
microsoft_data.reset_index(inplace = True)
microsoft_data.head()
```

## 9. Module IV: Document Object Model Scraping Pipeline (Asset B Fundamental Ingestion)

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/stock.html"
html_data_2 = requests.get(url).text

soup2 = BeautifulSoup(html_data_2, "html.parser")

microsoft_revenue = pd.DataFrame(columns = ["Date","Revenue"])
for row in soup2.find_all("tbody")[1].find_all("tr"):
    col = row.find_all("td")
    date = col[0].text
    rev = col[1].text
    microsoft_revenue = pd.concat([microsoft_revenue, pd.DataFrame({"Date" : [date], "Revenue" : [rev]})], ignore_index = True)

microsoft_revenue["Revenue"] = microsoft_revenue["Revenue"].str.replace(',|\$', "", regex=True)
microsoft_revenue.dropna(inplace=True)

microsoft_revenue.tail()
```

## 10. **Module V: Analytics Dashboard Matrix Implementation**

```python
make_graph(nvdia_data, nvdia_revenue, 'NVDIA')
```

```python
make_graph(microsoft_data, microsoft_revenue, 'Microsoft')
```
